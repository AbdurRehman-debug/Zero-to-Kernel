### Model Representation
as the linear regression model is a linear combination of the input features, we can represent it mathematically as:
$$
y = wx+b 

$$
now we will implement this model in Python.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1, 2, 3, 4, 5])
y_train = np.array([200,400,600,800,1000])
x_train.shape[0], y_train.shape[0]
```

### plot the data

```python
plt.scatter(x_train, y_train, marker='x', color='blue')
plt.title("Housing Prices")
# Set the y-axis label
plt.ylabel('Price (in 1000s of dollars)')
# Set the x-axis label
plt.xlabel('Size (1000 sqft)')
plt.show()
```

### take model paramaters

```python
w = 150
b = 130
```

### compute model output

```python
def model(x,w,b):
    f_wb = np.zeros(x.shape[0])
    for i in range(x.shape[0]):
        f_wb[i]  =w * x[i] + b
    return f_wb
```

### plot the model output


```python
plt.scatter(x_train, y_train)
plt.title("Housing Prices")
# Set the y-axis label
plt.ylabel('Price (in 1000s of dollars)')
# Set the x-axis label
plt.xlabel('Size (1000 sqft)')
plt.plot(x_train, model(x_train,w,b), color='r')
```
